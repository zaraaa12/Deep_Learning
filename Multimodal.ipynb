```python
from transformers import BlipProcessor, BlipForQuestionAnswering
import torch
import gradio as gr
from PIL import Image

# Muat prosesor dan model
processor = BlipProcessor.from_pretrained("Salesforce/blip-vqa-base")
model = BlipForQuestionAnswering.from_pretrained("Salesforce/blip-vqa-base")

# Atur perangkat
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

def vqa_answer(image, question):
    # Praproses input
    inputs = processor(image, question, return_tensors="pt").to(device)

    # Hasilkan jawaban
    with torch.no_grad():
        generated_ids = model.generate(**inputs)
        answer = processor.decode(generated_ids[0], skip_special_tokens=True)

    return answer

# Definisikan komponen input
image_input = gr.components.Image(type="pil", label="Unggah Gambar")
question_input = gr.components.Textbox(lines=1, placeholder="Masukkan pertanyaan Anda di sini...", label="Pertanyaan")

# Definisikan komponen output
answer_output = gr.components.Textbox(label="Jawaban")

# Buat antarmuka
iface = gr.Interface(
    fn=vqa_answer,
    inputs=[image_input, question_input],
    outputs=answer_output,
    title="Aplikasi Tanya Jawab Visual",
    description="Ajukan pertanyaan tentang gambar yang diunggah.",
    article="Aplikasi ini menggunakan model BLIP untuk menjawab pertanyaan tentang gambar."
)

# Luncurkan aplikasi
iface.launch()
```
